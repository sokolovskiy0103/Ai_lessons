### Питання 20
Маленький «round-trip»: серіалізація у рядок і назад.

```python
import json

original = {"score": 90, "tags": ["a", "b"]}

text = json.dumps(original)
restored = json.loads(text)

print(type(text).__name__)
print(restored == original)
print(restored["tags"][0])
```

✍️ *Ваша відповідь (що буде виведено і чому):*


<details>
<summary>💡 Відповідь</summary>

<pre><code>str
True
a</code></pre><code>json.dumps</code> повертає <b>рядок</b> (<code>str</code>). Після <code>json.loads</code> назад отримуємо рівнозначний словник, тож <code>restored == original</code> → <b>True</b>. <code>restored["tags"][0]</code> — перший елемент списку тегів, «<b>a</b>».

</details>
