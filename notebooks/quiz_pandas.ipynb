### Питання 20
Це спрощена логіка з гри «U.S. States»: за вгаданою назвою штату дістаємо його координати.

```python
import pandas as pd

data = pd.DataFrame({
    "state": ["California", "Texas", "Ohio"],
    "x": [-300, 50, 200],
    "y": [10, -40, 90],
})

guess = "Texas"
if guess in data["state"].values:
    row = data[data["state"] == guess]
    print(row["x"].values[0], row["y"].values[0])
else:
    print("Немає такого штату")
```

✍️ *Ваша відповідь (що буде виведено і чому):*


<details>
<summary>💡 Відповідь</summary>

<pre><code>50 -40</code></pre><code>guess in data["state"].values</code> перевіряє, чи є «Texas» серед назв — так. Фільтр <code>data[data["state"] == "Texas"]</code> дає рядок Техасу; <code>row["x"].values[0]</code> і <code>row["y"].values[0]</code> дістають <b>скалярні</b> координати <code>50</code> і <code>-40</code>. Саме за такими координатами гра ставить підпис штату на мапу.

</details>
