Чим відрізняються атрибути `name`, `_name` та `__name`? Який із них Python *реально* захищає, а який — лише «домовленість» між програмістами?

✍️ *Ваша відповідь:*


<details>
<summary>💡 Відповідь</summary>

<ul>
<li><code>name</code> — публічний атрибут, вільний доступ звідусіль.</li>
<li><code>_name</code> — «захищений» <b>лише за домовленістю</b>: одне підкреслення — сигнал «це внутрішнє, не чіпай ззовні», але технічно доступ нічим не обмежений.</li>
<li><code>__name</code> — Python застосовує <b>name mangling</b>: перейменовує атрибут на <code>_ClassName__name</code>, тому звичайне звертання <code>obj.__name</code> ззовні дає <code>AttributeError</code>.</li>
</ul>

Важливо: <b>справжнього</b> захисту немає ніде — навіть <code>__name</code> доступний через <code>obj._ClassName__name</code>. У Python приватність тримається на домовленостях, <code>__</code> лише ускладнює випадковий доступ.

</details>

</details>

✍️ *Ваша відповідь:*


<details>
<summary>💡 Відповідь</summary>

Геттери/сеттери дають <b>контроль над доступом</b>: можна валідувати нові значення (вік ≥ 0), обчислювати значення «на льоту», логувати звернення, змінити внутрішнє зберігання даних, не змінюючи зовнішній інтерфейс.

<code>@property</code> дозволяє користуватися цим контролем із <b>синтаксисом звичайного атрибута</b>: пишемо <code>p.age</code> та <code>p.age = 30</code> замість <code>p.get_age()</code> і <code>p.set_age(30)</code> — код коротший, а клас зберігає контроль. Бонус: можна почати з простого публічного атрибута, а пізніше перетворити його на property без зміни коду, який використовує клас.

</details>

### Питання 14
При створенні об'єкта програма «зависає» і падає з `RecursionError`. Знайдіть причину.

```python
class Person:
    def __init__(self, age):
        self.age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Вік не може бути від'ємним")
        self._age = value

p = Person(25)
print(p.age)
```

✍️ *У чому помилка?*


```python
# Виправлений код:
```

<details>
<summary>💡 Відповідь</summary>

Усередині сеттера рядок <code>self.age = value</code> — це знову присвоєння property <code>age</code>, тобто сеттер викликає <b>сам себе</b> → нескінченна рекурсія → <code>RecursionError</code>. Зберігати значення треба у «внутрішній» атрибут <code>self._age</code>:

<pre><code>    @age.setter
    def age(self, value):
        if value &lt; 0:
            raise ValueError("Вік не може бути від'ємним")
        self._age = value
</code></pre>

Зверніть увагу: <code>self.age = age</code> в <code>__init__</code> — це <b>не</b> помилка, а гарна практика: присвоєння проходить через сеттер, тож валідація спрацьовує вже при створенні об'єкта.

</details>

### Питання 20
Що станеться при виконанні цього коду? Якщo буде помилка — яка саме і чому? Як змінити клас, щоб `t.celsius = 30` запрацювало (з перевіркою, що температура не нижча за −273.15)?

```python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

t = Temperature(25)
t.celsius = 30
print(t.celsius)
```

✍️ *Ваша відповідь:*


```python
# Виправлений клас Temperature:
```

<details>
<summary>💡 Відповідь</summary>

Код впаде на рядку <code>t.celsius = 30</code> з <code>AttributeError: property 'celsius' of 'Temperature' object has no setter</code> (у старіших версіях Python — <code>can't set attribute</code>). Property, в якої означений лише геттер, доступна <b>тільки для читання</b>.

Щоб дозволити запис — додаємо сеттер з валідацією:

<pre><code>class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius   # проходить через сеттер → валідація

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value &lt; -273.15:
            raise ValueError("Температура нижча за абсолютний нуль")
        self._celsius = value
</code></pre>

</details>
