✍️ *Ваша відповідь:*


<details>
<summary>💡 Відповідь</summary>

<code>super()</code> дає доступ до методів <b>батьківського</b> класу. Найчастіше її використовують у <code>__init__</code> нащадка, щоб <b>повторно використати</b> ініціалізацію батька замість копіювання коду.

<pre><code>class Animal:

Якщо нащадок визначає власний <code>__init__</code> і <b>не</b> викликає <code>super().__init__()</code>, то код ініціалізації батька не виконається — атрибути батька (тут <code>name</code>) не з'являться.

</details>

✍️ *Ваша відповідь:*


<details>
<summary>💡 Відповідь</summary>

<b>has-a (композиція)</b>: об'єкт <b>містить</b> інший як частину. «Автомобіль <b>має</b> двигун» → у <code>Car</code> є атрибут <code>self.engine = Engine()</code>.

<pre><code>class Engine: ...

</code></pre>

Правило: якщо «X є Y» — успадкування; якщо «X має Y» — композиція. Часто композиція гнучкіша за глибокі ієрархії успадкування.

</details>

Метод `who()` визначено у трьох класах. Чий саме викличеться?

```python
class A:
    def who(self):
        return "A"


class B(A):
    def who(self):
        return "B"


class C(A):
    def who(self):
        return "C"


class D(B, C):
    pass


d = D()
print(d.who())
```

<summary>💡 Відповідь</summary>

<pre><code>B
</code></pre>

<code>D</code> сама не має <code>who()</code>, тож Python шукає метод за <b>MRO</b> (Method Resolution Order). Для <code>class D(B, C)</code> порядок такий: <code>D → B → C → A</code>. Перший знайдений <code>who()</code> — у <code>B</code>, тому виведе «B». (Подивитися порядок можна через <code>D.__mro__</code>.)

```python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        ...

    def __str__(self):
        return f"{type(self).__name__} площею {self.area()}"


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2


class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return round(3.14159 * self.r ** 2, 2)


for fig in [Square(4), Circle(1)]:
    print(fig)
```
