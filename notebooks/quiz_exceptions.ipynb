### Питання 20
Власний виняток у дії.

```python
class NotEnoughMoney(Exception):
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise NotEnoughMoney("Недостатньо коштів")
    return balance - amount

try:
    withdraw(100, 150)
except NotEnoughMoney as e:
    print("Помилка:", e)
```

✍️ *Ваша відповідь (що буде виведено і чому):*


<details>
<summary>💡 Відповідь</summary>

<pre><code>Помилка: Недостатньо коштів</code></pre>150 &gt; 100, тому <code>withdraw</code> піднімає власний виняток <code>NotEnoughMoney("Недостатньо коштів")</code>. Його ловить <code>except NotEnoughMoney as e</code>, а друк <code>e</code> показує <b>повідомлення</b>, передане в конструктор.

</details>
